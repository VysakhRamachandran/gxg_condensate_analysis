```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# -------------------------------
# Load data
# -------------------------------
df_3res = pd.read_csv('data/sasa_persidue_3res.csv')
df_9res = pd.read_csv('data/sasa_persidue_9res.csv')
df_9res_sp = pd.read_csv('data/sasa_persidue_9res_spacers.csv')

# -------------------------------
# Compute means
# -------------------------------
mean_3 = df_3res.mean()
mean_9 = df_9res.mean()
mean_9sp = df_9res_sp.mean()

# Combine all into one sequence
all_means = pd.concat([mean_3, mean_9, mean_9sp])
labels = [f"{s}\n3res" for s in mean_3.index] + \
         [f"{s}\n9res" for s in mean_9.index] + \
         [f"{s}\n9res_sp" for s in mean_9sp.index]

# -------------------------------
# Plot as single line
# -------------------------------
plt.figure(figsize=(12,5))
plt.bar(range(len(all_means)), all_means.values)
plt.xticks(range(len(all_means)), labels, rotation=45)
plt.ylabel('Average SASA')
plt.title('SASA per System and Dataset (Single Bar Line)')
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# -------------------------------
# Load data
# -------------------------------
df_3res_sasa = pd.read_csv('data/sasa_persidue_3res.csv')
df_9res_sasa = pd.read_csv('data/sasa_persidue_9res.csv')

# -------------------------------
# Compute statistics
# -------------------------------
means_3 = df_3res_sasa.mean()
std_3   = df_3res_sasa.std()

means_9 = df_9res_sasa.mean()
std_9   = df_9res_sasa.std()

systems = means_3.index
x = np.arange(len(systems))

# -------------------------------
# Create figure with 3 subplots
# -------------------------------
fig, axes = plt.subplots(1, 3, figsize=(18,5))

# ===============================
# 1. Side-by-side bar plot
# ===============================
width = 0.35

axes[0].bar(x - width/2, means_3.values, width,
            yerr=std_3.values, capsize=5, label='3res')

axes[0].bar(x + width/2, means_9.values, width,
            yerr=std_9.values, capsize=5, label='9res')

axes[0].set_xticks(x)
axes[0].set_xticklabels(systems, rotation=45)
axes[0].set_ylabel('Average SASA')
axes[0].set_title('Mean SASA Comparison')
axes[0].legend()

# ===============================
# 2. Boxplot comparison
# ===============================
data = []
labels = []

for col in systems:
    data.append(df_3res_sasa[col])
    labels.append(f"{col}\n3res")

    data.append(df_9res_sasa[col])
    labels.append(f"{col}\n9res")

axes[1].boxplot(data)
axes[1].set_xticks(range(1, len(labels)+1))
axes[1].set_xticklabels(labels, rotation=45)
axes[1].set_ylabel('SASA')
axes[1].set_title('Distribution Comparison')

# ===============================
# 3. Delta SASA plot
# ===============================
delta = means_9 - means_3

axes[2].bar(systems, delta.values)
axes[2].axhline(0, linestyle='--')

axes[2].set_ylabel('ΔSASA (9res - 3res)')
axes[2].set_title('Change with Chain Length')
axes[2].set_xticklabels(systems, rotation=45)

# -------------------------------
# Final layout
# -------------------------------
plt.tight_layout()
plt.show()
```

```python
stru = parmed.load_file('../gxg_comps/3_res/equal/output.top')

# or explicitly by atom names:
stru.strip('@OW,HW1,HW2')
u = mda.Universe(stru,
                 '../gxg_comps/3_res/equal/traj_2-nw.dcd')
```

```python
solute = u.select_atoms("protein")   # or "resname RNA" or "not resname SOL"

# Define transformations:
# 1. Center solute in box
# 2. Wrap everything around the solute
u.trajectory.add_transformations(mda.transformations.unwrap(u.select_atoms("all") ),
    center_in_box(solute),
    #wrap(solute, compound='fragments')
)
```
