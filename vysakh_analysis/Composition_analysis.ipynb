```python

import matplotlib.pyplot as plt
import matplotlib.patches as mpatches

# -----------------------------
# Residue groups
# -----------------------------
aromatic = ['PHE', 'TYR', 'TRP','HIS']
polar = ['CYS','THR','GLN','ASN','SER']
positive = ['ARG','LYS']
negative = ['GLU','ASP']
nonpolar = ['VAL','ILE','ALA','LEU','PRO','GLY','MET']

# Order residues by chemistry
ordered_residues = aromatic + polar + positive + negative + nonpolar
tot_df = tot_df.loc[ordered_residues]

# -----------------------------
# Colors for systems
# -----------------------------
colors = {
    'equal': 'grey',
    'charged': '#b30039',
    'aromatic': '#e3ba14',
    'polar': '#7aaa3d',
    'nonpolar': '#ac5cad'
}

# -----------------------------
# Plot
# -----------------------------
fig, ax = plt.subplots(figsize=(14,6))

tot_df.plot.bar(
    ax=ax,
    width=0.75,
    color=[colors[col] for col in tot_df.columns],
    edgecolor='none'
)

# -----------------------------
# Background shading for residue classes
# -----------------------------
sections = [
    (0, len(aromatic), '#e3ba14', 'Aromatic'),
    (len(aromatic), len(aromatic)+len(polar), '#7aaa3d', 'Polar'),
    (len(aromatic)+len(polar), len(aromatic)+len(polar)+len(positive), '#b30039', 'Positive'),
    (len(aromatic)+len(polar)+len(positive), len(aromatic)+len(polar)+len(positive)+len(negative), '#0c7ace', 'Negative'),
    (len(aromatic)+len(polar)+len(positive)+len(negative), len(ordered_residues), '#ac5cad', 'Nonpolar')
]

for start, end, color, label in sections:
    ax.axvspan(start-0.5, end-0.5, color=color, alpha=0.06)

# -----------------------------
# Vertical separators
# -----------------------------
boundaries = [
    len(aromatic),
    len(aromatic)+len(polar),
    len(aromatic)+len(polar)+len(positive),
    len(aromatic)+len(polar)+len(positive)+len(negative)
]

for b in boundaries:
    ax.axvline(b-0.5, color='black', linestyle='--', alpha=0.3)

# -----------------------------
# Labels & ticks
# -----------------------------
ax.set_xticklabels(ordered_residues, rotation=45, fontsize=12)
ax.set_ylabel('Clustered Ratio', fontsize=14)
ax.set_xlabel('Residue', fontsize=14)
ax.set_ylim(0.5, 1.0)

# -----------------------------
# Clean legend (systems only)
# -----------------------------
legend_handles = [
    mpatches.Patch(color=colors['equal'], label='Equal'),
    mpatches.Patch(color=colors['charged'], label='Charged'),
    mpatches.Patch(color=colors['aromatic'], label='Aromatic'),
    mpatches.Patch(color=colors['polar'], label='Polar'),
    mpatches.Patch(color=colors['nonpolar'], label='Nonpolar')
]

ax.legend(handles=legend_handles, loc='upper right', ncol=5, frameon=False)

# -----------------------------
# Final layout
# -----------------------------
plt.tight_layout()
plt.show()
```
