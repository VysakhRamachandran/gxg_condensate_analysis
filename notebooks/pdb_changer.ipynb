```python
def rechain_id(read_file, write_file):
    with open(read_file, 'r') as old:
        with open(write_file, 'w') as new:
            lines = old.readlines()
            chain = 0
            resid = 1
            for i, line in enumerate(lines):
                if i == 0:
                    newline = line
                    new.write(newline)
                elif i == 1:
                    newline = line
                    new.write(newline) 
                elif i == 2:
                    if line.split()[5] == '1':
                        newline = line.replace(' '+line.split()[4]+'   1', ' '+str(chain)+'   1')
                        new.write(newline)
                    elif line.split()[5] == '2':
                        newline = line.replace(' '+line.split()[4]+'   2', ' '+str(chain)+'   2')
                        new.write(newline)
                    else:
                        newline = line.replace(' '+line.split()[4]+'   3', ' '+str(chain)+'   3')
                        new.write(newline)
                else:
    #                 if len(lines[i].split()) == 5:
    #                     continue
                    if len(lines[i].split()) == 13:
                        if lines[i].split()[4] == lines[i-1].split()[4]:
                            if line.split()[5] == '1':
                                newline = line.replace(' '+line.split()[4]+'   1', ' '+str(chain)+'   1')
                                new.write(newline)
                            elif line.split()[5] == '2':
                                newline = line.replace(' '+line.split()[4]+'   2', ' '+str(chain)+'   2')
                                new.write(newline)
                            else:
                                line.split()[5] == '3'
                                newline = line.replace(' '+line.split()[4]+'   3', ' '+str(chain)+'   3')
                                new.write(newline)
                        elif lines[i].split()[4] != lines[i-1].split()[4]:
                            chain += 1
                            if line.split()[5] == '1':
                                newline = line.replace(' '+line.split()[4]+'   1', ' '+str(chain)+'   1')
                                new.write(newline)
                            elif line.split()[5] == '2':
                                newline = line.replace(' '+line.split()[4]+'   2', ' '+str(chain)+'   2')
                                new.write(newline)
                            else:
                                newline = line.replace(' '+line.split()[4]+'   3', ' '+str(chain)+'   3')
                                new.write(newline)
```
