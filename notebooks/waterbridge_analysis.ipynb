```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import gaussian_kde
import pandas as pd

# Example DataFrame (replace tot_wb_df with your actual DataFrame)
np.random.seed(0)
tot_wb_df = pd.DataFrame({
    'System A': np.random.normal(0, 1, 200),
    'System B': np.random.normal(2, 0.5, 200),
    'System C': np.random.normal(-1, 1.2, 200),
    'System D': np.random.normal(1, 0.8, 200)
})

fig, ax = plt.subplots(figsize=(7, 5))
positions = np.arange(1, tot_wb_df.shape[1] + 1)
max_width = 0.3  # Maximum half-width for each violin

for i, col in enumerate(tot_wb_df.columns):
    data = tot_wb_df[col].dropna().values
    # Compute the kernel density estimate for the data
    kde = gaussian_kde(data)
    # Evaluate the density only over the data range (cut=0 equivalent)
    x_eval = np.linspace(data.min(), data.max(), 100)
    density = kde(x_eval)
    # Normalize density so that the maximum width is max_width
    density = density / density.max() * max_width
    
    # Draw the violin shape using fill_betweenx
    ax.fill_betweenx(
        x_eval,
        positions[i] - density,
        positions[i] + density,
        alpha=0.7,
        label=col  # Optional: use for legend if desired
    )
    # Optionally, mark the median
    median = np.median(data)
    ax.plot([positions[i] - max_width, positions[i] + max_width],
            [median, median],
            color='k', linestyle='--', linewidth=1)

# Customize the plot
ax.set_xticks(positions)
ax.set_xticklabels(tot_wb_df.columns, rotation=45, fontsize=12)
ax.tick_params(axis='y', labelsize=12)
ax.set_xlabel('System', fontsize=14)
ax.set_ylabel('Clustered Residue Ratio', fontsize=14)

plt.title("Custom Violin Plot (Cut=Tails to Data Range)")
plt.show()
```

```python
def water_bridge_check(dens, replica, wb_object, df):

    res_list = ['ALA','ARG','ASN','ASP','CYS','GLN','GLU','GLY','HIS','ILE','LEU',
                'LYS','MET','PHE','PRO','SER','THR','TRP','TYR','VAL']

    water_bridge_df = pd.DataFrame(index=res_list, columns=res_list)
    water_bridge_df = water_bridge_df.fillna(float(0))
    
    top = 'good_output.top'
    
    if dens == 50:
        num_each = 5
        dcd = '4ms-traj.dcd'
        gly_u = mda.Universe(load_file(f'/Volumes/potoyan-lab/will/gxg-agg-project/{dens}mgml_100a/replica-{replica}/{top}'), 
                                f'/Volumes/potoyan-lab/will/gxg-agg-project/{dens}mgml_100a/replica-{replica}/{dcd}')

        gly = gly_u.select_atoms('resname GLY')

        selection = []
        for resid in gly.residues:
            if len(list(resid.atoms)) == 7:
                selection.append(resid.resnum)
                
    elif dens == 125:
        num_each = 15
        dcd = '5ms-traj.dcd'
        gly_u = mda.Universe(load_file(f'/Volumes/potoyan-lab/will/gxg-agg-project/{dens}mgml_100a/replica-{replica}/{top}'), 
                                f'/Volumes/potoyan-lab/will/gxg-agg-project/{dens}mgml_100a/replica-{replica}/{dcd}')

        gly = gly_u.select_atoms('resname GLY')

        selection = []
        for resid in gly.residues:
            if len(list(resid.atoms)) == 7:
                selection.append(resid.resnum)
                
    else:
        num_each = 30
        dcd = 'long-traj.dcd'
        dcd2 = 'long-traj2.dcd'
        
        gly_u = mda.Universe(load_file(f'/Volumes/potoyan-lab/will/gxg-agg-project/{dens}mgml_100a/replica-{replica}/{top}'), 
                                f'/Volumes/potoyan-lab/will/gxg-agg-project/{dens}mgml_100a/replica-{replica}/{dcd}',
                                    f'/Volumes/potoyan-lab/will/gxg-agg-project/{dens}mgml_100a/replica-{replica}/{dcd2}')

        gly = gly_u.select_atoms('resname GLY')

        selection = []
        for resid in gly.residues:
            if len(list(resid.atoms)) == 7:
                selection.append(resid.resnum)

    for residue in res_list:

        if residue == 'GLY':

            for res in res_list:

                value = 0

                frames = len(wb_object.frames)

                if res == 'GLY':

                    residue_res_df = df[df['sele1_resnm'].str.fullmatch(residue) & df['sele1_resid'].isin(selection) & df['sele2_resnm'].str.fullmatch(res) & df['sele2_resid'].isin(selection)]

                    value += (residue_res_df.shape[0]) / frames

                    water_bridge_df.loc[residue, res] = value
                    water_bridge_df.loc[res, residue] = value

                else:

                    residue_res_df = df[df['sele1_resnm'].str.fullmatch(residue) & df['sele1_resid'].isin(selection) & df['sele2_resnm'].str.fullmatch(res)]
                    res_residue_df = df[df['sele1_resnm'].str.fullmatch(res) & df['sele2_resnm'].str.fullmatch(residue) & df['sele2_resid'].isin(selection)]

                    value += (residue_res_df.shape[0] + res_residue_df.shape[0]) / frames

                    water_bridge_df.loc[residue, res] = value
                    water_bridge_df.loc[res, residue] = value

        else:

            for res in res_list:

                value = 0

                frames = len(wb_object.frames)    

                residue_res_df = df[df['sele1_resnm'].str.fullmatch(residue) & df['sele2_resnm'].str.fullmatch(res)]
                res_residue_df = df[df['sele1_resnm'].str.fullmatch(res) & df['sele2_resnm'].str.fullmatch(residue)]

                if residue == res:
                    value += residue_res_df.shape[0]
                else:
                    value += residue_res_df.shape[0] + res_residue_df.shape[0]

                value = value/frames

                if water_bridge_df.loc[residue, res] == 0.0:
                    water_bridge_df.loc[residue, res] = value

                if water_bridge_df.loc[res, residue] == 0.0:
                    water_bridge_df.loc[res, residue] = value

    water_bridge_df = water_bridge_df / num_each

    return water_bridge_df,residue_res_df,res_residue_df, selection

```
