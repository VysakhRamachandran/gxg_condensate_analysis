```python
import MDAnalysis as mda
import numpy as np
import pandas as pd
import solvation_analysis
from solvation_analysis.solute import Solute
```

```python
u = mda.Universe('/Volumes/potoyan-lab/will/gxg-agg-project/125mgml_100a/replica-1/output.pdb','/Volumes/potoyan-lab/will/gxg-agg-project/125mgml_100a/replica-1/5ms-traj.dcd')
```

```python
ala_sel = u.select_atoms('resname ALA and resnum 2 and element C')
arg_sel = u.select_atoms('resname ARG and resnum 2 and element C')
asn_sel = u.select_atoms('resname ASN and resnum 2 and element C')
asp_sel = u.select_atoms('resname ASP and resnum 2 and element C')
cys_sel = u.select_atoms('resname CYS and resnum 2 and element C')
gln_sel = u.select_atoms('resname GLN and resnum 2 and element C')
glu_sel = u.select_atoms('resname GLU and resnum 2 and element C')
gly_sel = u.select_atoms('resname GLY and resnum 2 and element C')
his_sel = u.select_atoms('resname HIS and resnum 2 and element C')
ile_sel = u.select_atoms('resname ILE and resnum 2 and element C')
leu_sel = u.select_atoms('resname LEU and resnum 2 and element C')
lys_sel = u.select_atoms('resname LYS and resnum 2 and element C')
met_sel = u.select_atoms('resname MET and resnum 2 and element C')
phe_sel = u.select_atoms('resname PHE and resnum 2 and element C')
pro_sel = u.select_atoms('resname PRO and resnum 2 and element C')
ser_sel = u.select_atoms('resname SER and resnum 2 and element C')
thr_sel = u.select_atoms('resname THR and resnum 2 and element C')
trp_sel = u.select_atoms('resname TRP and resnum 2 and element C')
tyr_sel = u.select_atoms('resname TYR and resnum 2 and element C')
val_sel = u.select_atoms('resname VAL and resnum 2 and element C')

solute_dict = {'ALA':ala_sel, 'ARG':arg_sel, 'ASN':asn_sel, 'ASP':asp_sel, 'CYS':cys_sel, 
               'GLN':gln_sel, 'GLU':glu_sel, 'GLY':gly_sel, 'HIS':his_sel, 'ILE':ile_sel,
               'LEU':leu_sel, 'LYS':lys_sel, 'MET':met_sel, 'PHE':phe_sel, 'PRO':pro_sel,
               'SER':ser_sel, 'THR':thr_sel, 'TRP':trp_sel, 'TYR':tyr_sel, 'VAL':val_sel}

solvent_sel = u.select_atoms('resname HOH')

radii_dict = {'ALA':10, 'ARG':10, 'ASN':10, 'ASP':10, 'CYS':10, 
               'GLN':10, 'GLU':10, 'GLY':10, 'HIS':10, 'ILE':10,
               'LEU':10, 'LYS':10, 'MET':10, 'PHE':10, 'PRO':10,
               'SER':10, 'THR':10, 'TRP':10, 'TYR':10, 'VAL':10}
```

```python
final_dict = {}
for i,j in solute_dict.items():
    num = len(j.atoms)/15
    solute = Solute.from_atoms(solvent_sel, {i:j}, radii=radii_dict, solute_name=i)
    solute.run(start=4000, stop=-1, step=50000)
    extra = solute.coordination.coordination_numbers
    for i,j in extra.items():
        extra[i] = j / num
    final_dict.update(extra)                                                              
```
