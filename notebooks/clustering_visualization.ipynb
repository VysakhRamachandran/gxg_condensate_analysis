```python
def load_data(dens=125, size=100, analysis_type='rdf', rep=1):
    
    '''Load rdf data
    dens: 50, 125, 250
    size: 80, 100 (length of sides of box in angstroms)
    analysis_type: rdf(cluster rdf), dist(cluster distances), cont(contact frequency), 
                   size(cluster size), comp(cluster composition), spatial(spatial cluster composition)
    rep: 1,2,3
    '''
    
    if analysis_type == 'surface':
        path = f'/Volumes/potoyan-lab/will/gxg-agg-project/{dens}mgml_{size}a/replica-{rep}/surface_comp.csv'
    else:
        path = f'/Volumes/potoyan-lab/will/gxg-agg-project/{dens}mgml_{size}a/replica-{rep}/cluster_{analysis_type}.csv'

#     if analysis_type =='rdf':
#         path = f'/Volumes/potoyan-lab/will/gxg-agg-project/{dens}mgml_{size}a/replica-{rep}/cluster_{analysis_type}.csv'
        
#     elif analysis_type =='cont':
#         path = f'/Volumes/potoyan-lab/will/gxg-agg-project/{dens}mgml_{size}a/replica-{rep}/ca_rdf{rdfnum}.csv'
        
#     elif analysis_type =='comp':
#         path = f'/Volumes/potoyan-lab/will/gxg-agg-project/{dens}mgml_{size}a/replica-{rep}/cluster_rdf_{cluster_type}_{stride}.csv'
    
#     elif analysis_type =='size':
#         path = f'/Volumes/potoyan-lab/will/gxg-agg-project/{dens}mgml_{size}a/replica-{rep}/ca_rdf{rdfnum}.csv'     
    
#     else:
#         path = f'/Volumes/potoyan-lab/will/gxg-agg-project/{dens}mgml_{size}a/replica-{rep}/side_rdf.csv'
    
    if analysis_type=='surface':
        return pd.read_csv(path, index_col=0)
    else:
        return pd.read_csv(path)
```

```python
def make_contact_frequency_dictionary(freq_files):
    '''
    go through a list of frequency files and record all of the frequencies for
    each replica
    '''
    contact_dictionary = {}
    regex = r'\w:\w+:\d+\s+\w:\w+:\d+'
    # go through each of the contact files and fill in the lists
    for i, file in enumerate(freq_files):
        with open(file, 'r') as freqs:
            for line in freqs.readlines():
                if re.search(regex, line):
                    line = line.strip()
                    first, second, num_str = line.split()
                    label = first + "-" + second
                    if label not in contact_dictionary.keys():
                        contact_dictionary[label] = [0 for n in range(i)]
                        contact_dictionary[label].append(float(num_str))
                    else:
                        contact_dictionary[label].append(float(num_str))
        #Extend all the lists before opening the next freq_file
        for key in contact_dictionary.keys():
            if i > 0 and len(contact_dictionary[key]) != i+1:
                length = len(contact_dictionary[key])
                extend = (i+1) - length
                contact_dictionary[key].extend([0 for n in range(extend)])
    return contact_dictionary

dens=50
rep=1

diction = make_contact_frequency_dictionary([f'../{dens}mgmL_100a/replica-{rep}/contact-freqs.tsv'])

newdict = {}

for key, value in diction.items():
    substring = 'ARG'
    if substring in key:
        string1 = str(key[2:5])
        string2 = str(key[9:13])
        newkey = string1+string2
        try:
            newdict[newkey].append(value)
        except KeyError:
            newdict[newkey] = [value]
    else:
        continue

for key, value in newdict.items():
    newdict[key] = np.mean(value)
    
newdict = {k: v for k, v in sorted(newdict.items(), key=lambda item: item[1], reverse=True)}

newdict
```

```python
from MDAnalysis.analysis import distances
from sklearn.cluster import DBSCAN
from collections import Counter
import MDAnalysis as mda
from parmed import load_file
from MDAnalysis.analysis.hydrogenbonds import HydrogenBondAnalysis

def clustering(u, frame, dens):

    ### Function that is nested in all of the analysis methods in this file ###

    ca_sel = u.select_atoms('resid 2 and name CA')

    cas = [atom.index for atom in ca_sel.atoms]

    to_cluster = [atom for i, atom in enumerate(u.trajectory[frame].positions)if i in cas]

    n_ca = len(ca_sel)

    self_distances = distances.self_distance_array(ca_sel.positions, box=u.trajectory[frame].dimensions)

    sq_dist_arr = np.zeros((n_ca, n_ca))
    triu = np.triu_indices_from(sq_dist_arr, k=1)

    sq_dist_arr[triu] = self_distances
    sq_dist_arr.T[triu] = self_distances

    if dens == 50:
        model = DBSCAN(eps=9, min_samples=1, metric='precomputed')
    elif dens == 125:
        model = DBSCAN(eps=9, min_samples=2, metric='precomputed')
    else:
        model = DBSCAN(eps=10, min_samples=2, metric='precomputed')

    labels = model.fit_predict(sq_dist_arr)
    b = Counter(labels)

    clustered_list = [atom for j, atom in enumerate(to_cluster) 
                      for i, item in enumerate(labels) 
                      if item == b.most_common()[0][0] 
                      and j == i]

    centroid = np.mean(clustered_list, axis=0)

    ind_list = [cas[index] for index, j in enumerate(to_cluster) 
                for i in clustered_list 
                if j[0]==i[0] and j[1]==i[1] and j[2]==i[2]]

    index_list = [x+1 for x in ind_list]

    res_list = [u.atoms[i].resindex for i in index_list]
    
    name_list = [u.atoms[i].resname for i in index_list]


    return centroid, to_cluster, clustered_list, res_list, name_list 
```

```python
def hist_cluster(univ, start, stop, step, prot_bins, water_bins):
        
    temp_prot_sel = univ.select_atoms('protein')
    temp_water_sel = univ.select_atoms('not protein and not name NA and not name CL')
    
    tp_count, tp_z = np.histogram(temp_prot_sel.positions[:,2], bins=prot_bins)
    tp_z = 0.5*(tp_z[1:]+ tp_z[:-1])
    
    tw_count, tw_z = np.histogram(temp_water_sel.positions[:,2], bins=water_bins)
    tw_z = 0.5*(tw_z[1:]+ tw_z[:-1])
    
    prot_df = pd.DataFrame(index = tp_z)
    water_df = pd.DataFrame(index = tw_z)
    
    for frame in univ.trajectory[start:stop:step]:
        
        sel_prot = univ.select_atoms('protein')
        sel_water = univ.select_atoms('not protein and not name NA and not name CL')

        prot_snap = sel_prot.positions[:,2]
        water_snap = sel_water.positions[:,2]

        prot_count, prot_z = np.histogram(prot_snap, bins=prot_bins, density=True)
        prot_z = 0.5*(prot_z[1:]+ prot_z[:-1])

        water_count, water_z = np.histogram(water_snap, bins=water_bins, density=True)
        water_z = 0.5*(water_z[1:]+ water_z[:-1])

        mean_index = np.argmin(np.abs(np.array(prot_count)-np.mean(prot_count)))

        if mean_index > len(z)//2:
            prot_count = np.roll(prot_count, mean_index)
            water_count = np.roll(water_count, mean_index)

        else:
            prot_count = np.roll(prot_count, -mean_index)
            water_count = np.roll(water_count, -mean_index)
            
        frame_num = frame.frame
            
        prot_df[f'{frame_num}'] = prot_count
        water_df[f'{frame_num}'] = water_count
        
    final_prot = pd.DataFrame(prot_df.mean(axis=1), columns = ['NumberDensity'])
    final_water = pd.DataFrame(water_df.mean(axis=1), columns = ['NumberDensity'])
        
    return final_prot, final_water
    
```
