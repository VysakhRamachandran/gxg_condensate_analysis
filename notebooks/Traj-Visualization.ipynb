```python
# Load the topology file
topology = load_file('/Volumes/potoyan-lab/will/hnrnpa1/75mgml/output.top')

# Step 2: Define the new residue indices for protein residues only
protein_residues = [residue for residue in topology.residues if residue.name not in {"SOL", "WAT", "NA", "CL"}]

#Step 3
original_indices = np.arange(len(protein_residues))  # Number of protein residues
cycle_length = 45
start_value = 0  # Change to 1 if you want indices to start from 1
new_indices = original_indices % cycle_length + start_value

# Step 4: Apply new residue indices to protein residues
for residue, new_index in zip(protein_residues, new_indices):
    residue_number = new_index  # Assign the new residue index
    for atom in residue.atoms:
        atom.residue.number = residue_number  # Update residue number for all atoms in the residue

# Step 5: Save the modified topology file
topology.write('/Volumes/potoyan-lab/will/hnrnpa1/75mgml/reindexed_top.top')
```
