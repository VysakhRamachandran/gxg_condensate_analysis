# Make Interaction Count .csv file

```python
u = mda.Universe(f'/Volumes/potoyan-lab/will/rna_gxg/data/ALA/output.gro',
                     f'/Volumes/potoyan-lab/will/rna_gxg/data/ALA/trajectory.dcd')
```

```python
res_list = ['ALA','ARG','ASN','ASP','CYS','GLN','GLU','GLY','HIS','ILE',
            'LEU','LYS','MET','PHE','PRO','SER','THR','TRP','TYR','VAL']

int_list = ['VdWContact','Hydrophobic','HBDonor','Cationic','HBAcceptor','CationPi']

calc_start = 4000
calc_step = 10

for comp in ['charged','aromatic','polar','nonpolar']:
    u = mda.Universe(f'/Volumes/potoyan-lab/will/gxg-agg-project/125mgml_100a/compositions/{comp}/output-nw.gro',
                     f'/Volumes/potoyan-lab/will/gxg-agg-project/125mgml_100a/compositions/{comp}/7ms_traj-nw.dcd')
    
    res_dict = {}
    for res in res_list:
        sel = u.select_atoms(f'resid 2 and resname {res}')
        n_res = sel.n_residues
        res_dict[res] = n_res
    
    n_frames = (u.trajectory.n_frames - calc_start) / calc_step
    df = pd.DataFrame(columns = res_list, index = int_list)
    for res in res_list:
        temp_df = pd.read_csv(f'/Volumes/potoyan-lab/will/gxg-agg-project/125mgml_100a/compositions/analysis_data/{comp}_plf_files/{res}_plf.csv', index_col=0)
        count_dict = Counter(temp_df.iloc[1])
        for i,j in count_dict.items():
            df.loc[i, res] = j
    df = df.fillna(0)
    
    for column in df.columns:
        for i,j in res_dict.items():
            if i == column:
                df[column] /= j
    
    df = df / n_frames
    df.to_csv(f'/Volumes/potoyan-lab/will/gxg-agg-project/125mgml_100a/compositions/analysis_data/int_count_data/{comp}_int_count.csv')
```
