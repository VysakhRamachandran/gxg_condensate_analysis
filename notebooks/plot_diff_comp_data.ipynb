```python
aromatic = ['PHE', 'TYR', 'TRP']
polar = ['CYS','THR','GLN','ASN','SER']
nonpolar = ['VAL','ILE','ALA','LEU','PRO','GLY','MET']
positive = ['ARG','LYS','HIS']
negative = ['GLU','ASP']

res_list = ['ALA','ARG','ASN','ASP','CYS','GLN','GLU','GLY','HIS','ILE',
            'LEU','LYS','MET','PHE','PRO','SER','THR','TRP','TYR','VAL']

### Droplet Composition Alterations ###
for i in ['c_comp','a_comp','p_comp','n_comp']:

    if i == 'c_comp':
        df = c_comp
        group = positive + negative
        big_num = 23
        low_num = 11
    elif i == 'a_comp':
        df = a_comp
        group = aromatic
        big_num = 26
        low_num = 13
    elif i == 'p_comp':
        df = p_comp
        group = polar
        big_num = 23
        low_num = 11
    else:
        df = n_comp
        group = nonpolar
        big_num = 22
        low_num = 11
        
    for row in df[df['Residue'].isin(group)].index.tolist():
        df.loc[row, 'Frequency'] = df.loc[row, 'Frequency'] * 15 / big_num
    for row in [i for i in df.index.tolist() if i not in 
                df[df['Residue'].isin(group)].index.tolist()]:
        df.loc[row, 'Frequency'] = df.loc[row, 'Frequency'] * 15 / low_num
        
    df = df.sort_values(by='Frequency')
        
    if i == 'c_comp':
        c_comp = df
    elif i == 'a_comp':
        a_comp = df
    elif i == 'p_comp':
        p_comp = df
    else:
        n_comp = df

### Droplet Size Alterations###
c_size['Frame'] = c_size['Frame'] / 25
a_size['Frame'] = a_size['Frame'] / 25
p_size['Frame'] = p_size['Frame'] / 25
n_size['Frame'] = n_size['Frame'] / 25

```

```python
from matplotlib.legend_handler import HandlerBase

fig, ax = plt.subplots(figsize=[7,5])

aromatic = ['PHE', 'TYR', 'TRP']
polar = ['CYS','THR','GLN','ASN','SER']
nonpolar = ['VAL','ILE','ALA','LEU','PRO','GLY','MET']
positive = ['ARG','LYS','HIS']
negative = ['GLU','ASP']

charged_colors = ["#e60049" if j in positive else '#e6d800' if j in aromatic else '#87bc45' if j in polar 
                          else '#0d88e6' if j in negative else '#bd7ebe' for j in c_comp['Residue']]

aromatic_colors = ["#e60049" if j in positive else '#e6d800' if j in aromatic else '#87bc45' if j in polar 
                          else '#0d88e6' if j in negative else '#bd7ebe' for j in a_comp['Residue']]

polar_colors = ["#e60049" if j in positive else '#e6d800' if j in aromatic else '#87bc45' if j in polar 
                          else '#0d88e6' if j in negative else '#bd7ebe' for j in p_comp['Residue']]

nonpolar_colors = ["#e60049" if j in positive else '#e6d800' if j in aromatic else '#87bc45' if j in polar 
                          else '#0d88e6' if j in negative else '#bd7ebe' for j in n_comp['Residue']]

ax.scatter(c_comp['Residue'], c_comp['Frequency'], color=charged_colors, marker='D')

ax.scatter(a_comp['Residue'], a_comp['Frequency'], color=aromatic_colors, marker='s')

ax.scatter(p_comp['Residue'], p_comp['Frequency'], color=polar_colors, marker='o')

ax.scatter(n_comp['Residue'], n_comp['Frequency'], color=nonpolar_colors, marker='x')



list_color  = ["black", "black", "black", "black"]
list_mak    = ["D","s","o","x"]
list_lab    = ["Charged", "Aromatic", "Polar", "Nonpolar"]

class MarkerHandler(HandlerBase):
    def create_artists(self, legend, tup, xdescent, ydescent,
                        width, height, fontsize, trans):
        return [plt.Line2D([width/2], [height/2.],ls="",
                       marker=tup[1],color=tup[0], transform=trans)]

ax.legend(list(zip(list_color,list_mak)), list_lab,
          bbox_to_anchor=(0.226, 1.01),
          handler_map={tuple:MarkerHandler()})

plt.xticks(rotation=45);

```
